Exercise 4: Implementing Classes and Objects for a Bookstore
You are tasked with creating a Python program that represents books and performs various
operations on the bookstore inventory.
Tasks

2: Add Methods for Displaying Book Information
• Add a method display_details() to the Book class that prints the details of the

3: Create a Bookstore Class
• Define a new class named Bookstore.
• Add the following:
---> An attribute inventory (a list to store Book objects).
---> An __init__ method to initialize the inventory.
4: Add Methods to Manage the Inventory
• Add the following methods to the Bookstore class:
---> add_book(book): Add a Book object to the inventory.
---> display_inventory(): Display details of all books in the inventory.
5: Add Search and Remove Features
• Add the following methods to the Bookstore class:
---> search_book(title): Search for a book by title and display its details if
found.
---> remove_book(title): Remove a book from the inventory by its title.

#### SOLUTION FOR EXERCISE-4 IS GIVEN BELLOW ####

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_details(self):
        print(f"Title: {self.title}, Author: {self.author}, Price: ${self.price}")
class Bookstore:
    def __init__(self):
        self.inventory = []

    def add_book(self, book):
        self.inventory.append(book)

    def display_inventory(self):
        for book in self.inventory:
            book.display_details()

    def search_book(self, title):
        for book in self.inventory:
            if book.title == title:
                book.display_details()
                return
        print("------>>>>> Book not found <<<<<------")

    def remove_book(self, title):
        for book in self.inventory:
            if book.title == title:
                self.inventory.remove(book)
                print(f"Removed book: {title}")
                return
        print("------>>>>> Book not found <<<<<------")
```
