**Exercise-3: The following Python code declares the Node class, and constructs a tree using the list-of-children data structure.**

image.png

i.	Write code to access the node with value 'g'.

ii.	Find root.children[0].value.


**Exercise -3 : Solution**


```python
class Node:
    def __init__(self, value, children=None):
        self.value = value
        self.children = children or []

    def __repr__(self):
        if self.children:
            fmt = 'Node({!r}, {!r})'
        else:
            fmt = 'Node({!r})'
        return fmt.format(self.value, self.children)

# Define the tree
root = Node('a', [
    Node('b', [Node('c')]),
    Node('d'),
    Node('e', [Node('f'), Node('g')])
])

# i. Access the node with value 'g'
node_g = root.children[2].children[1]
print(node_g)  # Output: Node('g')

# ii. Find root.children[0].value
value_b = root.children[0].value
print(value_b)  # Output: 'b'
```
