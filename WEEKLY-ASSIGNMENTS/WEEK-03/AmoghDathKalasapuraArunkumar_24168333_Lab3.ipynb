**Exercise-1.c: Compare the memory layout of the two structures from Exercise 1.a and Exercise 1.b. Explain the differences in terms of memory allocation, memory usage, and pointer usage.**

**Solution 1.c:** solution
There are Three main differences are there in exercise 1a and Exercise 1b 

1. Memory Allocation:

Char Array : The char array is stored in contiguous memory locations. Where each element/item occupies a specific memory address sequentially.

Linked List : The linked list nodes are not necessarily stored in contiguous memory locations. Each node contains a value and a pointer to the next node's memory address

2. Memory Usage:

Char Array: Uses less memory as it stores only the values in contiguous memory locations.

Linked List: Uses more memory because each node includes both the value and the pointer to the next node.

3. Pointer Usage:

Char Array: A single pointer is required to point to the base address of the array.

Linked List: Each node contains a pointer to the next node, which means more pointers are used, and traversal requires following the pointers.

In short ,a char array offers simplicity in memory allocation and usage, while a linked list provides flexibility in dynamic memory allocation at the cost of additional memory for pointers and potential fragmentation.

```python
class Node:
    def __init__(self, data):
        self.data = data  # Store data
        self.next = None  # Pointer to next node
```

*Linked-List*

```python
class LinkedList:
    def __init__(self):
        self.head = None  # Initialize an empty list

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node

    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Delete node from front
    def delete_front(self):
        if not self.head:  # If list is empty
            print("List is empty, nothing to delete")
            return
        self.head = self.head.next  # Move head to next node

    # Delete node from end
    def delete_end(self):
        if not self.head:  # If list is empty
            print("List is empty, nothing to delete")
            return
        if not self.head.next:  # If only one node is present
            self.head = None
            return
        temp = self.head
        while temp.next.next:  # Traverse till second last node
            temp = temp.next
        temp.next = None  # Remove last node

    def delete_node(self, key):
        temp = self.head

        if temp and temp.data == key:  # If head is the key
            self.head = temp.next
            temp = None
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            return

        prev.next = temp.next
        temp = None

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")
```
