a) Write necessary Python code that creates the vertices and edges of the graph.

```python
#solution for Exercise 1a:
class Node:
    def __init__(self, value):
        self.value = value
        self.neighbours = []

    def add_neighbour(self, node):
        self.neighbours.append(node)

# Create vertices
node_zero  = Node(0)
node_one   = Node(1)
node_two   = Node(2)
node_three = Node(3)
node_four  = Node(4)

# Add edges
node_zero.add_neighbour(node_one)
node_zero.add_neighbour(node_two)
node_zero.add_neighbour(node_three)

node_one.add_neighbour(node_zero)
node_one.add_neighbour(node_two)
node_one.add_neighbour(node_three)
node_one.add_neighbour(node_four)

node_two.add_neighbour(node_zero)
node_two.add_neighbour(node_one)
node_two.add_neighbour(node_three)

node_three.add_neighbour(node_zero)
node_three.add_neighbour(node_one)
node_three.add_neighbour(node_two)
node_three.add_neighbour(node_four)

node_four.add_neighbour(node_one)
node_four.add_neighbour(node_three)

# Function to print nodes and their neighbours
def print_graph(nodes):
    for node in nodes:
        neighbours_values = [neighbour.value for neighbour in node.neighbours]
        print(f"Node {node.value} -> {neighbours_values}")

# List of all nodes
nodes = [node_zero, node_one, node_two, node_three, node_four]

# Print the graph
print_graph(nodes)
```
