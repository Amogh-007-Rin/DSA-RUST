### Adaptation of evaluate_infix function using stack class:

```python
class Stack:

    def __init__(self):
        self.stack = []
    
    def is_empty(self):
        return len(self.stack) == 0
    
    def push(self,item):
        self.stack.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return "stack is empty"
    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            return "stack is empty"
        
def precedence(op):
    """Returns the precedence of an operator."""
    if op == '^':
        return 3
    if op in ('*', '/', '%'):
        return 2
    if op in ('+', '-'):
        return 1
    return 0

def apply_operator(operands, operator):
    # Applies an operator to the top two values in the operand stack
    b = operands.pop()  # Second operand (right side)
    a = operands.pop()  # First operand (left side)

    if operator == '+':
        return a + b
    if operator == '-':
        return a - b
    if operator == '*':
        return a * b
    if operator == '/':
        return a / b
    if operator == '^':
        return a ** b

def evaluate_infix(expression):
    ## evaluation of infix expression using two stacks ##
    operand_stack = Stack()
    operator_stack = Stack()

    tokens = expression.split() ## This will tokenize it to the elements
    
    for token in tokens:
        if token.isdigit(): 
            operand_stack.push(int(token))
        elif token == '(': 
            operator_stack.push(token)
        elif token == ')':  
            while not operator_stack.is_empty() and operator_stack.peek() != '(':
                res = apply_operator(operand_stack, operator_stack.pop())
                operand_stack.push(res)
            operator_stack.pop()  # This will Remove '(' from the existing stack
        else:  # Operator (+, -, *, /, ^) are involved
            while (not operator_stack.is_empty() and precedence(operator_stack.peek()) >= precedence(token)
                   and operator_stack.peek() != '('):
                res = apply_operator(operand_stack, operator_stack.pop())
                operand_stack.push(res)
            operator_stack.push(token)

    while not operator_stack.is_empty():  # This is used to Process remaining operators
        res = apply_operator(operand_stack, operator_stack.pop())
        operand_stack.push(res)

    return operand_stack.pop()

##### Driver code / Implementation ####

expr = "( 3 + 5 ) * 2 ^ 2 - 10 / 2 - 2 + 3 * 2"
result = evaluate_infix(expr)
print("Result:", result) ### In this case the result is 31.0
```

# Exercise 4: Using the provided Stack and Queue implementations, write a Python function to
# check if a given sequence is a palindrome by leveraging both data structures together.
# Example of Palindrome: 1221, 550055, 1111, 41814, huh, yay, etc.
# [ Palindrome - reads the same backwards as forwards]
# [Hint: Insert each element on the Stack and Queue. Then compare values returned by
# dequeue and pop operations. If all match, it is a palindrome, otherwise it is not]

### solution for exercise 4 ###


```python
class Stack:

    def __init__(self):
        self.stack = []
    
    def is_empty(self):
        return len(self.stack) == 0
    
    def push(self,item):
        self.stack.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return "stack is empty"
    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            return "stack is empty"
        
class Queue:
    def __init__(self):
        self.queue = []

    def is_empty(self):
        return len(self.queue) == 0

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.queue.pop(0)
        else:
            return "Queue is empty"

    def peek(self):
        if not self.is_empty():
            return self.queue[0]
        else:
            return "Queue is empty"

### Creating a function to categorize palindromes ###

def is_palindrome(sequence):
    stack = Stack()
    queue = Queue()

    # Adding each element to both stack and queue 
    for char in sequence:
        stack.push(char)
        queue.enqueue(char)

    # Compare each element of stack and queue with each other
    while not stack.is_empty() and not queue.is_empty():
        if stack.pop() != queue.dequeue():
            return False

    return True

# Driver code / Implimentation
sequences = ["1221", "550055", "1111", "41814", "huh", "yay", "hello"]
for seq in sequences:
 print(f"{seq}: {is_palindrome(seq)}")
 
```
